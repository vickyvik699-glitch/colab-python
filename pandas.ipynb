```python
import pandas as pd
data = [['devendra',10],['nikhil',20],['raj',30],['hello',40]]
df = pd.DataFrame(data)
print(df)

# add cols name
df = pd.DataFrame(data,columns=['name','age'])
print(df)

# add indexing name
df = pd.DataFrame(data,index=['a','b','c','d'])
print(df)
```
